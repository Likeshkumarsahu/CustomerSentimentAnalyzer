```python
import torch
import numpy as np
from transformers import DistilBertTokenizerFast, DistilBertForSequenceClassification
```

```python
model_path = "sentiment_distilbert_best"

tokenizer = DistilBertTokenizerFast.from_pretrained(model_path)
model = DistilBertForSequenceClassification.from_pretrained(model_path)

model.eval()
```

```python
text="""I have used this product for some time, and the overall experience has been average. The product functions as expected and does not have any major flaws, but it also does not offer anything particularly impressive. The build quality is reasonable for the price, and performance is mostly consistent.

There are a few features that are useful, while others could have been improved. I did not encounter any serious issues during use. Overall, the product meets basic requirements, and whether it is worth buying depends on individual expectations."""
```

```python
inputs = tokenizer(
    text,
    return_tensors="pt",
    truncation=True,
    max_length=256
)
```

```python
with torch.no_grad():
    outputs = model(**inputs)

logits = outputs.logits
probs = torch.softmax(logits, dim=1)

pred_class = probs.argmax().item()
confidence = probs.max().item()

pred_class, confidence
```
